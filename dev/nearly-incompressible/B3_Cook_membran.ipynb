# Cook-membrán — volumetrikus locking
Borsodi Bálint TDK-dolgozata, B3 (7.2. táblázat). Bal él befogva, a jobb élre függőlegesen felfelé ható, egységnyi összerő jut. Az elmozdulást a terhelt él közepén mérjük. A két merevség az LLFEM DSL-jével készül.

A négy hálóméret és két Poisson-tényező eredményei.

```julia
cook(x,y) = (48x, 44x + (44-28x)*y)
ref = [16.543 18.751 5.208 15.909;
       20.094 20.979 5.352 17.649;
       21.404 21.668 5.791 18.211;
       21.685 21.810 6.404 18.343]
println("n      FI (ν=0,3)  SRI (ν=0,3)  FI (ν=0,4999)  SRI (ν=0,4999)")
for (i, n) in enumerate((4, 8, 16, 24))
    U, xy = q4_grid(n, n; map=cook)
    bc = [BoundaryCondition("left"; field=U, ux=0, uy=0)]
    f = ∫(U ⋅ [0.0, 1/16]; Γ="right")
    tip = argmin([hypot(x-48, y-52) for (x,y) in eachcol(xy)])
    values = [solveField(K, f; support=bc).a[2tip, 1]
              for ν in (0.30, 0.4999) for K in q4_stiffness(U, 1.0, ν)]
    @assert maximum(abs, values - ref[i,:]) < 0.0006
    @printf("%2d       %7.3f       %7.3f       %7.3f          %7.3f\n", n, values...)
end
```
