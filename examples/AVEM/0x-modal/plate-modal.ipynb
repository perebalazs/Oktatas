```julia
using LowLevelFEM, Arpack
```

```julia
structured_rect_mesh(lx=10)
```

```julia
openPreProcessor()
```

```julia
mat = Material("body", E=2e5, ν=0.3, ρ=7.85e-9)
U = Field([mat], type=:VectorField, dim=2, fieldName=:u);
```

```julia
K = ∫(ε(U) ⋅ D(:PlaneStress, mat) ⋅ ε(U))
K[:,:]
```

```julia
M = ∫(U ⋅ U * mat.ρ)
M[:,:]
```

```julia
bc = BoundaryCondition("left", ux=0, uy=0);
```

```julia
u = solveEigen(K, M, normalize=:unit);
```

```julia
getEigenValues(u)
```

```julia
showDoFResults(u, name="eigen free")
```

```julia
u2 = solveEigen(K, M, support=[bc]);
```

```julia
showDoFResults(u2, name="eigen fixed");
```

```julia
f = ∫(U ⋅ [1,0], Γ="right");
```

```julia
u3 = solveField(K - 2 * M, f, support=[bc])
```

```julia
showDoFResults(u3, name="exited u")
```
