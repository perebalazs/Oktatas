```julia
using LowLevelFEM
```

```julia
openGeometry("lemet1.geo")
#openPreProcessor()
```

```julia
mat = Material("plate", E=2e5, ν=0.3)
P = Problem([mat], type=:PlaneStress)
```

```julia
bc = displacementConstraint("left", ux=0, uy=0)

ld = load("right", fx=1)
```

```julia
u = solveDisplacement(P, support=[bc], load=[ld])
```

```julia
showDoFResults(u)
```

```julia
s = solveStress(u)
showDoFResults(s)
```
