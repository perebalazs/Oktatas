```julia
mat = material("body", E=2e5, ν=0.3)
prob = Problem([mat], type=:Solid)

load1 = load("right", fy=-1)
supp = displacementConstraint("left", ux=0, uy=0, uz=0)

u = solveDisplacement(prob, load=[load1], support=[supp])
σ = solveStress(u)

showDoFResults(u, name="u", visible=true, factor=100)
showStressResults(σ, :x, name="σx")
```
