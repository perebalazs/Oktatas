```julia
using LowLevelFEM
```

```julia
openGeometry("tarto.geo")
```

```julia
mat = material("part", E=2e5, ν=0.3)
prob = Problem([mat]);
```

```julia
supp1 = displacementConstraint("base", uz=0)
supp2 = displacementConstraint("hole1", ux=0, uy=0)
supp3 = displacementConstraint("perim", uz=-0.00001)

load1 = load("hole2", fy=-5);
```

```julia
u = solveDisplacement(prob, load=[load1], support=[supp1, supp2, supp3])
showDoFResults(u, name="u", visible=true, factor=1);
```

```julia
ε = solveStrain(u)
showStrainResults(ε, name="ε");
```

```julia
σ = solveStress(u)
showStressResults(σ, name="σ");
```
