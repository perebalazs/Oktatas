```julia
using LowLevelFEM
```

```julia
openGeometry("grad-beam.geo")
#openPreProcessor()
```

```julia
mat = Material("line")
U = Field([mat], type=:VectorField, dim=3, fieldName=:u)
```

```julia
t = tangentVector(U, "line")
showElementResults(t, name="t")
n = VectorField([-t[2], t[1], t[3]])
showElementResults(n, name="n")
```

```julia
s1 = ScalarField(U, "line", (x,y,z)->0.1 * x^2)
s2 = ScalarField(U, "line", (x,y,z)->-0.1 * y)
s3 = ScalarField(U, "line", (x,y,z)->0.0 * z)

u = VectorField([s1,s2,s3])
```

```julia
showDoFResults(u, name="u", factor=1)
```

```julia
grad_u_t = ∇(u) * t

showElementResults(grad_u_t, name="grad t")
```

```julia
showElementResults(t * grad_u_t, name="ε")
showElementResults(n * grad_u_t, name="γ")
```
