```julia
structured_box_mesh(x0=-10, y0=-10, z0=-10, lx=20, ly=20, lz=20)

mat = Material("body")
P = Problem([mat]);
```

```julia
ex = VectorField(P, "body", [1, 0, 0])
ey = VectorField(P, "body", [0, 1, 0]);
```

```julia
probe(ex ∘ ex, 0,0,0)
```

```julia
probe(ex ∘ ey, 0,0,0)
```

```julia
α = deg2rad(30)

a = VectorField(P, "body", [cos(α), sin(α), 0])
b = VectorField(P, "body", [-sin(α), cos(α), 0])

v = VectorField(P, "body", [2, 1, 0])

T = a ∘ ex + b ∘ ey
probe(T, 0,0,0)
```

```julia
w = T * v
probe( w, 0,0,0)
```
