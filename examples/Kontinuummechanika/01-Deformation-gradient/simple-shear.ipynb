```julia
using LowLevelFEM
```

```julia
openGeometry("cube.geo")
```

```julia
mat = material("body")
problem = Problem([mat])
```

```julia
γ = 0.5
x(X, Y, Z) = X + γ * Y
y(X, Y, Z) = Y
z(X, Y, Z) = Z
```

```julia
r = vectorField(problem, "body", [x,y,z])
```

```julia
showDeformationResults(r, :vector, visible=true)
```

```julia
openPostProcessor()
```

```julia
#F = ∇(r)
F = r ∘ ∇
```

```julia
atP = "P"
FatP = probe(F, atP)
```

```julia
R = nodePositionVector(problem)

showDeformationResults(F * R, :vector)
```

```julia
#r = nodesToElements(r)
showDeformationResults(inv(F) * r, :vector)
```
